### Task 7: Feature and Target Separation

The dataset was divided into input features (X) and target variable (y).

The input dataset contains 23 features, while the target variable is `placement_status`.

The `student_id` and `salary_package_lpa` columns were excluded before feature separation. The target variable was kept separate to prevent target leakage and ensure realistic model training.

The resulting input shape is (100000, 23), and the target shape is (100000,).

```python

from sklearn.model_selection import train_test_split

# Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display shapes
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python

print("Training Data Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Data Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python

# Identify numerical and categorical features
numerical_features = X_train.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=['str', 'object']
).columns.tolist()

print("Numerical Features:", numerical_features)
print("\nCategorical Features:", categorical_features)
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Create preprocessing transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'),
         categorical_features)
    ]
)

# Fit on training data and transform
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data
X_test_processed = preprocessor.transform(X_test)

# Convert target into numerical format
y_train_encoded = y_train.map({
    'Placed': 1,
    'Not Placed': 0
})

y_test_encoded = y_test.map({
    'Placed': 1,
    'Not Placed': 0
})

print("Training Data Shape:", X_train_processed.shape)
print("Testing Data Shape:", X_test_processed.shape)
print("Preprocessing Completed!")
```

```python

print("Mean of first 5 scaled numerical features:")
print(X_train_processed[:, :5].mean(axis=0))

print("\nStandard deviation:")
print(X_train_processed[:, :5].std(axis=0))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("All ML Models Imported Successfully!")
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.neighbors import KNeighborsClassifier

models = {
    'Logistic Regression': LogisticRegression(
        max_iter=500, solver='lbfgs'
    ),
    'Random Forest': RandomForestClassifier(
        n_estimators=50,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ),
    'SVM': LinearSVC(
        C=1.0,
        max_iter=1000,
        random_state=42
    ),
    'KNN': KNeighborsClassifier(
        n_neighbors=5,
        n_jobs=-1
    )
}

trained_models = {}

for name, model in models.items():
    print("Training:", name)
    model.fit(X_train_processed, y_train_encoded)
    trained_models[name] = model

print("\nAll Models Trained Successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

results = []

predictions = {}

for name, model in trained_models.items():

    y_pred = model.predict(X_test_processed)
    predictions[name] = y_pred

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test_encoded, y_pred),
        'Precision': precision_score(y_test_encoded, y_pred),
        'Recall': recall_score(y_test_encoded, y_pred),
        'F1 Score': f1_score(y_test_encoded, y_pred)
    })

results_df = pd.DataFrame(results)

print(results_df)
```

Four machine learning classification models were trained:

1. Logistic Regression

```python

from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier

rf_params = {
    'n_estimators': [50, 100],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    param_distributions=rf_params,
    n_iter=5,
    cv=3,
    scoring='f1',
    random_state=42,
    n_jobs=-1
)

rf_search.fit(X_train_processed, y_train_encoded)

print("Best Parameters:", rf_search.best_params_)
print("Best Cross Validation F1 Score:", rf_search.best_score_)
```

### Hyperparameter Tuning and Model Improvement

RandomizedSearchCV was applied to the Random Forest classifier using 3-fold cross-validation and F1 Score as the evaluation metric.

The best hyperparameters obtained were:

- Number of estimators: 50
- Maximum depth: 10
- Minimum samples split: 10
- Minimum samples leaf: 1

The best cross-validation F1 Score was **0.68135 (68.14%)**.

**Conclusion:** Hyperparameter tuning helped identify a better configuration for the Random Forest model. Its performance will be verified on the unseen test dataset.

```python

best_rf = rf_search.best_estimator_

best_rf_pred = best_rf.predict(X_test_processed)

print("Tuned Random Forest Test Results")
print("Accuracy:", accuracy_score(y_test_encoded, best_rf_pred))
print("Precision:", precision_score(y_test_encoded, best_rf_pred))
print("Recall:", recall_score(y_test_encoded, best_rf_pred))
print("F1 Score:", f1_score(y_test_encoded, best_rf_pred))

# Save tuned model
trained_models['Tuned Random Forest'] = best_rf
```

```python

def predict_placement(student_data):

    # Convert input into DataFrame
    input_df = pd.DataFrame([student_data])

    # Apply the same preprocessing
    input_processed = preprocessor.transform(input_df)

    # Predict placement
    prediction = best_model.predict(input_processed)[0]

    # Get placement probability
    probability = best_model.predict_proba(input_processed)[0][1]

    # Risk classification
    if probability >= 0.70:
        risk = "Low Risk"
    elif probability >= 0.40:
        risk = "Medium Risk"
    else:
        risk = "High Risk"

    print("Placement Prediction:",
          "Placed" if prediction == 1 else "Not Placed")

    print("Placement Probability:", round(probability * 100, 2), "%")
    print("Risk Category:", risk)


print("Student Placement Prediction System Created!")
```
